```python
import pandas as pd
import json 

file_path = 'combined_df_filtered_by_images_20К_queries_deep_10_labeled_by_qwen3_vl_copy.json'
df = pd.read_json(file_path, lines=True)
df.head()
```

```python
df['latency_total_s'].mean()
```

```python
df[
    (df['response_text'].isna()) &
    (df['error'].notna())
]['error'].values[0]
```

```python
err_counter = 0

def iter_load(x):
    global err_counter
    try:
        return json.loads(x)
    except json.JSONDecodeError as e:
        print(f'err: {err_counter}')
        err_counter += 1
        return {}
        raise

df.loc[
    (df['response_text'].isna()) &
    (df['error'].notna()), # rows where answer is empty with an error present 
    "response_text"
] = "{}"
df["response_text"] = df["response_text"].apply(iter_load)
df["label"] = df["response_text"].apply(lambda x: x.get('класс') if 'класс' in x else x.get('Класс'))
df["label"].value_counts()
```

```python
df = df[df["label"].notna()] 
df["label"] = df["label"].astype(int)
df["label"].value_counts()
```

```python
df[
    ['q_item_id', 'd_item_id', 'label']
].to_csv('combined_df_filtered_by_images_20К_queries_deep_10_qwen_labels_out.csv', index=False)
```
